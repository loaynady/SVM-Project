SVM Model

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.feature_extraction.text import TfidfVectorizer

data= {
    'v1': [
        'ham','spam','ham','spam','ham','spam','ham','spam','ham','spam',
        'ham','spam','ham','spam','ham','spam','ham','spam','ham','spam'
    ],
    'v2': [
        'Hey how are you doing today',
        'Congratulations you won a free ticket call now',
        'Are we still meeting later',
        'Win cash prize click this link now',
        'I will call you in a minute',
        'Limited offer buy now and get discount',
        'Did you finish your homework',
        'You have been selected for a reward claim now',
        'Lets go to the gym tomorrow',
        'Free entry in a contest text now',
        'Can you send me the file please',
        'Get cheap meds online limited time offer',
        'See you at the meeting',
        'You won a lottery call this number',
        'What are you doing tonight',
        'Exclusive deal just for you click now',
        'Dont forget the appointment',
        'Act now to receive your prize',
        'Thanks for your help',
        'You are a lucky winner claim your reward'
    ]
}

df= pd.DataFrame(data)
df
```

```python
df["label"]=df["v1"].map({"spam":1,"ham":0})
df
```

```python
x=df["v2"]
y=df["label"]
```

```python
vectorizer=TfidfVectorizer(stop_words="english")
x_vectorized=vectorizer.fit_transform(x)

x_reduced=x_vectorized[:,:2].toarray()
x_reduced
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_reduced,y,test_size=0.2,random_state=42)
```

```python
kernels=["linear","poly"]
models={}

for kernel in kernels:
    model=SVC(kernel=kernel,random_state=42)
    model.fit(x_train,y_train)
    models[kernel]=model

for kernel,model in models.items():
    y_pred=model.predict(x_test)
    accuracy=accuracy_score(y_test,y_pred)
    print(f"accuracy for kernel {kernel} : {accuracy:.4f}")
```
